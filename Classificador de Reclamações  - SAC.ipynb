```python
from transformers import pipeline

# Carrega o modelo de classificação (1 a 5 estrelas)
classificador = pipeline(
    "text-classification",
    model="nlptown/bert-base-multilingual-uncased-sentiment"
)

print("=== CLASSIFICADOR DE RECLAMAÇÕES DO IFOOD ===")
print("Digite 'sair' para encerrar.")
print("---------------------------------------------")

while True:
    texto = input("Digite a reclamação: ").title().strip()

    if texto.lower() == "sair":
        print("Encerrado.")
        break

    if texto == "":
        print("Digite algo válido!\n")
        continue

    # Faz a classificação
    resultado = classificador(texto)[0]

    # O modelo retorna algo tipo "3 stars"
    estrelas = int(resultado["label"][0])
    score = resultado["score"]

    # Converte estrelas → categoria SAC
    if estrelas == 1:
        categoria = "⚠️ Problema Grave"
    elif estrelas == 2:
        categoria = "⚠️ Reclamação"
    elif estrelas == 3:
        categoria = "😐 Neutro"
    elif estrelas == 4:
        categoria = "😊 Satisfeito"
    else:
        categoria = "🤩 Muito satisfeito / Elogio"

    print("\n====== RESULTADO ======")
    print(f"Texto: {texto}")
    print(f"Estrelas: {estrelas} ⭐")
    print(f"Categoria: {categoria}")
    print(f"Confiança: {score:.2f}")
    print("=======================\n")
```
